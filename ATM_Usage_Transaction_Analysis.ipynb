```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries Imported Successfully!")
```

```python

data = {
    'Transaction_ID': ['T001', 'T002', 'T003', 'T004', 'T005',
                       'T006', 'T007', 'T008', 'T009', 'T010'],

    'ATM_ID': ['ATM01', 'ATM02', 'ATM01', 'ATM03', 'ATM02',
               'ATM01', 'ATM03', 'ATM02', 'ATM01', 'ATM03'],

    'Location': ['Salem', 'Erode', 'Salem', 'Coimbatore', 'Erode',
                 'Salem', 'Coimbatore', 'Erode', 'Salem', 'Coimbatore'],

    'Date': ['2026-11-01', '2026-11-01', '2026-11-02',
             '2026-11-02', '2026-11-03', '2026-11-03',
             '2026-11-04', '2026-11-04', '2026-11-05',
             '2026-11-05'],

    'Time': ['08:15', '09:30', '10:45', '13:20', '18:30',
             '11:00', '15:30', '17:45', '20:00', '12:15'],

    'Transaction_Type': ['Withdrawal', 'Deposit', 'Withdrawal',
                         'Balance Inquiry', 'Withdrawal',
                         'Deposit', 'Withdrawal', 'Withdrawal',
                         'Balance Inquiry', 'Deposit'],

    'Amount': [2000, 5000, 3000, 0, 5000,
               4000, 2000, 3000, 0, 6000],

    'Account_Type': ['Savings', 'Current', 'Savings', 'Savings',
                     'Current', 'Savings', 'Current', 'Savings',
                     'Savings', 'Current'],

    'Status': ['Success', 'Success', 'Success', 'Success',
               'Failed', 'Success', 'Success', 'Success',
               'Success', 'Failed']
}

df = pd.DataFrame(data)

print(df)
```

```python

total_transactions = len(df)

print("Total ATM Transactions:", total_transactions)
```

```python

transaction_count = df['Transaction_Type'].value_counts()

print(transaction_count)
```

```python

status_count = df['Status'].value_counts()

print(status_count)
```

```python

average_amount = df['Amount'].mean()

print("Average Transaction Amount:", average_amount)
```

```python

maximum_amount = df['Amount'].max()

print("Maximum Transaction Amount:", maximum_amount)
```

```python

minimum_amount = df['Amount'].min()

print("Minimum Transaction Amount:", minimum_amount)
```

```python

most_used = df['Transaction_Type'].mode()[0]

print("Most Used Transaction Type:", most_used)
```

```python

location_count = df['Location'].value_counts()

print("ATM Usage by Location:")
print(location_count)
```

```python

atm_count = df['ATM_ID'].value_counts()

print("ATM Machine Usage:")
print(atm_count)
```

```python

plt.figure(figsize=(8,5))

transaction_count.plot(kind='bar', color='skyblue')

plt.title('ATM Transaction Type Analysis')
plt.xlabel('Transaction Type')
plt.ylabel('Number of Transactions')

plt.xticks(rotation=0)
plt.show()
```

```python

plt.figure(figsize=(6,6))

plt.pie(status_count,
        labels=status_count.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title('ATM Transaction Status Analysis')

plt.show()
```

```python

plt.figure(figsize=(8,5))

plt.hist(df['Amount'], bins=5, edgecolor='black')

plt.title('Transaction Amount Distribution')
plt.xlabel('Transaction Amount')
plt.ylabel('Frequency')

plt.show()
```

```python

plt.figure(figsize=(8,5))

location_count.plot(kind='bar', color='orange')

plt.title('ATM Usage by Location')
plt.xlabel('Location')
plt.ylabel('Number of Transactions')

plt.xticks(rotation=0)
plt.show()
```

```python

plt.figure(figsize=(8,5))

atm_count.plot(kind='bar', color='green')

plt.title('ATM Machine Usage Analysis')
plt.xlabel('ATM ID')
plt.ylabel('Number of Transactions')

plt.xticks(rotation=0)
plt.show()
```

```python

df['Hour'] = pd.to_datetime(
    df['Time'].astype(str),
    format='%H:%M:%S'
).dt.hour

print(df[['Time', 'Hour']])
```

```python

hourly_usage = df['Hour'].value_counts().sort_index()

print(hourly_usage)
```

```python

busiest_hour = hourly_usage.idxmax()

print("Busiest ATM Hour:", busiest_hour)
```

```python

plt.figure(figsize=(10,5))

hourly_usage.plot(kind='line', marker='o', color='red')

plt.title('ATM Usage by Hour')
plt.xlabel('Hour of the Day')
plt.ylabel('Number of Transactions')

plt.grid(True)
plt.show()
```

```python

daily_usage = df['Date'].value_counts().sort_index()

print("Daily ATM Transactions:")
print(daily_usage)
```

```python

plt.figure(figsize=(10,5))

daily_usage.plot(kind='bar', color='purple')

plt.title('Daily ATM Transaction Analysis')
plt.xlabel('Date')
plt.ylabel('Number of Transactions')

plt.xticks(rotation=45)
plt.show()
```

```python

busiest_day = daily_usage.idxmax()

print("Busiest Transaction Day:", busiest_day)
print("Number of Transactions:", daily_usage.max())
```

```python

amount_by_type = df.groupby('Transaction_Type')['Amount'].sum()

print("Total Amount by Transaction Type:")
print(amount_by_type)
```

```python

success_count = (df['Status'] == 'Success').sum()

total_count = len(df)

success_rate = (success_count / total_count) * 100

print("Transaction Success Rate:", success_rate, "%")
```

```python

plt.figure(figsize=(6,5))

plt.bar(['Success Rate', 'Failure Rate'],
        [success_rate, 100-success_rate],
        color=['green', 'red'])

plt.title('ATM Transaction Success Rate')
plt.ylabel('Percentage')

plt.show()
```

```python

print("TOTAL ATM TRANSACTIONS:", len(df))

print("TOTAL TRANSACTION AMOUNT:", df['Amount'].sum())

print("AVERAGE TRANSACTION AMOUNT:", df['Amount'].mean())

print("SUCCESSFUL TRANSACTIONS:",
      (df['Status'] == 'Success').sum())

print("FAILED TRANSACTIONS:",
      (df['Status'] == 'Failed').sum())

print("MOST USED TRANSACTION:",
      df['Transaction_Type'].mode()[0])

print("MOST USED ATM:",
      df['ATM_ID'].mode()[0])
```

```python

plt.figure(figsize=(8,5))

sns.barplot(data=df, x='Transaction_Type',
            y='Amount', estimator='mean')

plt.title('Average Amount by Transaction Type')
plt.xlabel('Transaction Type')
plt.ylabel('Average Amount (Rs.)')

plt.show()
```

```python

account_usage = df['Account_Type'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(account_usage,
        labels=account_usage.index,
        autopct='%1.1f%%')

plt.title('ATM Usage by Account Type')

plt.show()
```

```python

plt.figure(figsize=(8,5))

sns.countplot(data=df, x='Status')

plt.title('ATM Transaction Status')
plt.xlabel('Transaction Status')
plt.ylabel('Transaction Count')

plt.show()
```
